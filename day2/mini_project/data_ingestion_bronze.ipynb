```python
from pyspark.sql.functions import *
def add_ingestion(df):
    df1=df.withColumn("ingestion_date",current_date())
    return df1
```

```python

dbutils.widgets.text("catalog","dev")
dbutils.widgets.text("schema","amitb10_bronze")
dbutils.widgets.text("account_table","account_info")
dbutils.widgets.text("cust_profile_table","customer_prfile")
dbutils.widgets.text("transactions_table","transactions")

catalog=dbutils.widgets.get("catalog")
schema=dbutils.widgets.get("schema")
account_table=dbutils.widgets.get("account_table")
cust_profile_table=dbutils.widgets.get("cust_profile_table")
transactions_table=dbutils.widgets.get("transactions_table")
```

```python
input_path="/Volumes/dev/naval/raw"
```

```python
# transactions_table 
transactions_df = spark.read.csv(f"{input_path}/{transactions_table}.csv",header=True,inferSchema=True)
```

```python
transactions_df.show()
```

```python
final_transactions_df=add_ingestion(transactions_df)
```

```python
final_transactions_df.write.mode("overwrite").saveAsTable(f"{catalog}.{schema}.{transactions_table}")
```

```python
df=spark.read.json(f"{input_path}/{account_table}.json")
df.write.mode("overwrite").saveAsTable(f"{catalog}.{schema}.{account_table}")
```

```python
df=spark.read.json(f"{input_path}/{cust_profile_table}.json")
df.write.mode("overwrite").saveAsTable(f"{catalog}.{schema}.{cust_profile_table}")
```
